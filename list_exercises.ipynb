# List Exercises

### List Manipulation

```python
tools = ['Laptop', 'Monitor', 'Keyboard', 'Mouse']
print(tools[0])
print(tools[-1]) #grabs item from end of the list

sub_list = tools[1:3] # this grabs the fist key and stops at the index listed (noninclusive)
print(sub_list)
```

```python
# Modifying Elements
tools[1] = 'Webcam'
print(tools)
```

```python
# Add Elements
tools.append('Desk')
print(tools)
```

```python
tools.insert(1, 'Headphones')
print(tools)
```

```python
# Remove Elements
tools.remove('Keyboard')
print(tools)
```
