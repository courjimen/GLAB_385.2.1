# Reference
- creating a copy of a list DOESN'T create a new list. It points to the same place in memory
- Google: Pass by value vs Pass vs Reference
- Changing one referenced list, will also change the other

```python
og_list = [1, 2, 3, 4, 5]

new_list = og_list
print('OG', og_list)
print('New', new_list)

new_list.append(6)
print('OG', og_list)
print('New', new_list)
```

```python
#deep copy - a separate copy of a non-prim data type (not just a ref)
#shallow copy - a semi separate copy of a non-prim data type
second_copy = og_list.copy()
print(second_copy)
print(og_list)

og_list.append(7)
print(second_copy)
print(og_list)
```

```python
# Reverse, extends, sort

num_list = [1, 2, 3, 4, 5]
print(num_list)

num_list.reverse()
print(num_list)

num_list.extend([6, 7, 8, 9]) #adds multiple values to list (append only allows one value)
print(num_list)

num_list.sort()
print(num_list)
```

```python
# Membership Operators 
print(1 in num_list) #returns boolean T/F great for conditionals if statements
print(1 not in num_list)

if 1 in num_list:
    print("You're #1 🎉")
```

```python
# Create a func. Take in 2 lists of years, reverse chronological order, add them together
    # Funtions create reusuable codeblocks, can be reused as many times as needed (use "def" to define/declare a func)
    # def INSERT FUNC NAME(insert parameters/input): define what you're doing --> return (output value)
# defined function 'reorder_profit_years'  
# takes 2 parameters (years1, years2)
def reorder_profit_years(years1, years2): 
    # reversed order of years
    years1.reverse()
    years2.reverse()

    # added lists together
    years1.extend(years2)

    #returned newly extended list
    return years1
```

```python
year_list1 = [2026, 2025, 2024, 2023]
year_list2 = [1990, 1989, 1988, 1987]

print(reorder_profit_years(year_list2, year_list1))
```
