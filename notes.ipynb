### Steps to Create Env & ipynb (python notebook)
1.  Create project folder
2.  Open project folder in VSC
3.  Verify your explorer tab and terminal are in the right place
    *** NO NESTED REPOSITORIES OR PROJECTS ***
4.  (optional) initialize repo if pushing to github
5.  To create virtual env
    - in terminal run: `python3 -m venv .venv`
    *** If it fails, delete and run again ***
6.  Create `.gitignore` file at the ROOT of the project (next to the .venv, NOT IN)
7.  Document `.venv/` inside `.gitignore`
8.  Activate env
    -   In terminal: `source .venv/Scripts/activate`
    -   In terminal: `source .venv/bin/activate`
8.  Install ipykernel
    -   In terminal: `pip install ipykernel`
9.  Create `.ipynb` file
10. Select Kernel (top right of ipynb file once open)
    -   select the env you created (should be reccommended with a star)

#### Data Collections
- lists (arrays) - indexed collection. Have order and stay in order. Numbered by index.
- dictionaries (dict) - objects - keyed collection. Have no order, use a key to access information.
- tuples - records (rows of a table)

### String

- Collection of single characters.
- Data access within lists and string is IDENTICAL
- Index numbers to access values

```python
greet = 'Hello, World!'

# print the first value in a string/list
print(greet[0])
print(greet[1])
print(greet[2])
```

#### Python Indexing operator
- can be used with strings and lists
- `[ start : stop : step ]`
- by default, if we dont include the step, its 1

```python
print(greet[0:5])
print(greet[::-1]) # Goes backwards
print(greet[0::2]) # Goes to the end skipping every other letter
```

#### Type () function
- tells us what data type a variable is
- `type(value) -> data type`

```python
print(type(greet))
print(type(13))
print(type(True))
print(type(13.2))
```

### Typecasting Functions (to change the datatype of)
-   `str()` - turns anything into a string
-   `float()` - turns value into a float if possible
-   `int()` - turns value into a int if possible

```python
age = '30'
print(type(age))

# If you permenantly want to change a variable you have to re-assign it
age = int(age)

print(type(age))
```

```python
print(type(age))
```

#### len() - returns the length of a string or list
-   `len(arg) -> int`
-   works with:
    -   strings
    -   lists
    -   dictionaries
    -   tuples

```python
print(len(greet))
```

#### Methods - built in functions to python objects (everything in python is technically an object), for different datatypes/objects/modules
-   functions that are connected to varibles of every type
-   defined with a period at the end of a variable name, followed by the method
-   end with parenthases to invoke (run, execute) the method (function)

```python
print(greet.upper())
print(greet.split())
```

## Lists - indexed collections
-   Numbered indexes
-   Start counting at 0
-   Any data type can be in a list
-   A list can be made of multiple different data types
-   No-fixed length (mutable - it can be changed)
-   syntax - `[ ]` with comma separated values

```python
people = ['Carolyn', 'Joe', 'Dylan', 'Akorede']
```

```python
print(people[0:])
print(len(people))
```

### Types of Error:
- `IndexError` - out of range
- `TypeError` - wrong type in a list index, wrong type in an operation
- `SyntaxError` - can't understand code structure, incorrect code grammer
- `NameError` - Variable/function is not defined
- `ValueError` - Correct type, invalid value
- `KeyError` - dictionary key that doesn't exist
- `AttibuteError` - Object doesn't have the attribute/method you are trying to use
